# CESR planar wiggler：从 four-potential 到 transport map

这个 notebook 推导 CESR 中两个周期型 planar wiggler 的连续场模型，并展示如何用 SciBmad 的 FourPotentialParams 与隐式 Yoshida 辛积分器自动产生粒子传输和一阶 transport map。

目标不是把 Tao 输出的矩阵写死，而是建立参数化链条

$$
(B_{\max}, L_{\rm period}, N_{\rm period})
\longrightarrow A^\mu(x,y,s,t)
\longrightarrow \text{symplectic tracking}
\longrightarrow (c,R,T,\ldots).
$$

Tao/Bmad 的 map 在这里是验证基准，而不是最终模型本身。

## 1. Bmad planar-model 场

Bmad 周期型 planar wiggler 场为

$$
\begin{aligned}
B_x &= -B_{\max}\frac{k_x}{k_y}\sin(k_xx)\sinh(k_yy)\cos\theta,\\
B_y &=  B_{\max}\cos(k_xx)\cosh(k_yy)\cos\theta,\\
B_s &= -B_{\max}\frac{k_s}{k_y}\cos(k_xx)\sinh(k_yy)\sin\theta,
\end{aligned}
$$

$$
k_y^2=k_x^2+k_s^2,\qquad k_s=\frac{2\pi}{L_{\rm period}},\qquad
\theta=k_ss+\phi_s.
$$

CESR lattice 只给出 B_MAX、L_PERIOD 和 N_PERIOD。对应简化模型取 $k_x=0$，所以 $k_y=k_s\equiv k_w$：

$$
B_x=0,\qquad
B_y=B_{\max}\cosh(k_wy)\cos\theta,\qquad
B_s=-B_{\max}\sinh(k_wy)\sin\theta.
$$

Bmad 使用 $\phi_s=-k_wL/2$，使轴上竖直场关于元素中心对称。

## 2. 从磁场构造 four-potential

$$
A^\mu=(\phi,A_x,A_y,A_s),\qquad
\mathbf E=-\nabla\phi-\frac{\partial\mathbf A}{\partial t},\qquad
\mathbf B=\nabla\times\mathbf A.
$$

势有 gauge freedom。对静磁 planar wiggler，选择

$$
\phi=A_y=A_s=0,\qquad
A_x=\frac{B_{\max}}{k_w}\cosh(k_wy)\sin(k_ws+\phi_s).
$$

于是

$$
B_y=\frac{\partial A_x}{\partial s}
=B_{\max}\cosh(k_wy)\cos\theta,
$$

$$
B_s=-\frac{\partial A_x}{\partial y}
=-B_{\max}\sinh(k_wy)\sin\theta.
$$

使用势的关键优点是 SciBmad 可以从相应 Hamiltonian 进行辛积分。

## 3. Transport map 的定义

令 canonical 坐标为 $\mathbf z=(x,p_x,y,p_y,z,p_z)^T$。非线性 map 在参考轨道附近展开为

$$
z_i^{\rm out}=c_i+\sum_jR_{ij}z_j^{\rm in}
+\frac{1}{2}\sum_{jk}T_{ijk}z_j^{\rm in}z_k^{\rm in}+\cdots,
$$

$$
c=\mathcal M(0),\qquad
R_{ij}=\left.\frac{\partial\mathcal M_i}{\partial z_j}\right|_{z=0}.
$$

无辐射 Hamiltonian tracking 的一阶矩阵应满足

$$
R^TJR=J,\qquad
J=\operatorname{diag}\!\left(
\begin{bmatrix}0&1\\-1&0\end{bmatrix},
\begin{bmatrix}0&1\\-1&0\end{bmatrix},
\begin{bmatrix}0&1\\-1&0\end{bmatrix}\right).
$$

下面用有限差分显示 $R$；正式高阶工作流可用 GTPSA 从同一个 tracking function 产生 map。

```julia
using SciBmad
using FiniteDiff
using LinearAlgebra
using CairoMakie

cesr_dir = if isfile(joinpath(@__DIR__, "wiggler.jl"))
    @__DIR__
else
    joinpath(@__DIR__, "lattices", "cesr")
end
include(joinpath(cesr_dir, "wiggler.jl"))
using .WigglerModels
```

## 4. Beamlines four_potential 接口

planar_wiggler_four_potential(x,y,s,t,params) 返回：

1. (phi, Ax, Ay, As)；
2. 四个势分量分别对 (x,y,s,t) 的 16 个偏导数。

本模型返回实际磁势，因此元素设置 four_potential_normalized=false。BeamTracking 会用 beamline 的 p_over_q_ref 自动归一化，所以同一个 element 可以自然适用于不同参考能量和电荷符号。

构造器使用 Yoshida(order=6, n_steps=N_period*slices_per_period)。CESR 默认是 $12\times16=192$ 个 integration steps。

```julia
const CESR_B_MAX = 1.17               # T
const CESR_L_PERIOD = 0.19625         # m
const CESR_N_PERIOD = 12
const CESR_L = CESR_N_PERIOD * CESR_L_PERIOD
const CESR_P0C = 5.2889999753148e9    # eV

scales = planar_wiggler_scales(
    B_max=CESR_B_MAX,
    L_period=CESR_L_PERIOD,
    p0c=CESR_P0C,
)

scales
```

这些量提供直接的 sanity checks：

$$
B\rho\simeq17.6422\ {\rm T\,m},\qquad
\rho_w=\frac{B\rho}{B_{\max}}\simeq15.0788\ {\rm m}.
$$

$$
a_x=\frac{1}{\rho_wk_w^2}\simeq64.698\ \mu{\rm m},\qquad
a_{x'}=\frac{1}{\rho_wk_w}\simeq2.071\ {\rm mrad}.
$$

$a_x$ 复现原 CESR 文件中的 OSC_AMPLITUDE。平均竖直聚焦为

$$
\langle K_y\rangle\simeq\frac{1}{2\rho_w^2}
\simeq2.1991\times10^{-3}\ {\rm m}^{-2}.
$$

```julia
k_w = 2pi / CESR_L_PERIOD
phase = -k_w * CESR_L / 2
params = (CESR_B_MAX, k_w, phase)

s_grid = range(0, CESR_L; length=1201)
B_y_axis = [planar_wiggler_field(0.0, 0.0, s, params)[2] for s in s_grid]

fig = Figure(size=(900, 360))
ax = Axis(fig[1, 1], xlabel="local s [m]", ylabel="B_y(x=0,y=0) [T]",
          title="CESR planar-wiggler on-axis field")
lines!(ax, s_grid, B_y_axis, linewidth=2)
fig
```

```julia
x_test, y_test, s_test, t_test = 1.0e-3, 2.0e-3, 0.37, 0.0
potential, dA = planar_wiggler_four_potential(
    x_test, y_test, s_test, t_test, params,
)

# With Ay = As = 0: By = dAx/ds and Bs = -dAx/dy.
B_from_curl = (0.0, dA[7], -dA[6])
B_direct = planar_wiggler_field(x_test, y_test, s_test, params)

(B_from_curl=B_from_curl,
 B_direct=B_direct,
 max_error=maximum(abs.(B_from_curl .- B_direct)))
```

## 5. 构造并跟踪 Wiggler element

PlanarWiggler 返回普通 Beamlines LineElement，但附有 FourPotentialParams 和 Yoshida tracking method。物理输入仍然是 B_max 和周期参数，而不是冻结后的矩阵。

```julia
wiggler = PlanarWiggler(
    name="WIG_TEST",
    L=CESR_L,
    B_max=CESR_B_MAX,
    L_period=CESR_L_PERIOD,
    N_period=CESR_N_PERIOD,
    slices_per_period=16,
    order=6,
)

wiggler_line = Beamline(
    [wiggler],
    species_ref=Species("electron"),
    E_ref=CESR_P0C,
)

wiggler
```

```julia
tracked = track(
    wiggler_line;
    v0=zeros(1, 6),
    use_KA=false,
    use_explicit_SIMD=false,
)

zeroth_order = vec(tracked.v[1, :, end])
```

CESR wiggler 含整数个周期，因此轴上粒子的出口横向位置和角度应接近零。第五个坐标仍可非零：粒子在 wiggler 内摆动，路径比参考直线稍长。这是用普通 drift 替代时会丢失的效应。

```julia
function one_wiggler_map(v)
    result = track(
        wiggler_line;
        v0=reshape(v, 1, 6),
        use_KA=false,
        use_explicit_SIMD=false,
    )
    return vec(result.v[1, :, end])
end

R = FiniteDiff.finite_difference_jacobian(one_wiggler_map, zeros(6))
J2 = [0.0 1.0; -1.0 0.0]
J = kron(Matrix{Float64}(I, 3, 3), J2)
symplectic_residual = opnorm(transpose(R) * J * R - J, Inf)

(R=R, det_R=det(R), symplectic_residual=symplectic_residual)
```

## 6. 与平均聚焦近似比较

忽略周期内快速运动后，水平面近似为 drift，竖直面近似为常聚焦通道：

$$
R_x\approx\begin{bmatrix}1&L\\0&1\end{bmatrix},
$$

$$
R_y\approx
\begin{bmatrix}
\cos(\sqrt{K_y}L)&\sin(\sqrt{K_y}L)/\sqrt{K_y}\\
-\sqrt{K_y}\sin(\sqrt{K_y}L)&\cos(\sqrt{K_y}L)
\end{bmatrix}.
$$

这个近似只用于检查 full-field map 的低阶行为，不替代连续场。

```julia
K_y = scales.K_y_average
root_K = sqrt(K_y)
R_x_average = [1.0 CESR_L; 0.0 1.0]
R_y_average = [
    cos(root_K * CESR_L)              sin(root_K * CESR_L) / root_K
   -root_K * sin(root_K * CESR_L)     cos(root_K * CESR_L)
]

(horizontal_numeric=R[1:2, 1:2],
 horizontal_average=R_x_average,
 vertical_numeric=R[3:4, 3:4],
 vertical_average=R_y_average)
```

## 7. Tao/Bmad 验证与 CESR 集成

Linux/Bmad 环境中可用下面的命令查看零阶和一阶 map：

    tao -noinit -noplot -lat /absolute/path/to/cesr.bmad
    Tao> show ele WIG_W -xfer_mat
    Tao> show ele WIG_E -xfer_mat

建议比较 zeroth-order vector、全部 6×6 entries、path length、平均竖直聚焦、off-momentum map，以及最终 CESR tunes/Twiss。

需要区分两类基准：Bmad runge_kutta/连续场积分适合验证 four-potential；当前 CESR 默认的 bmad_standard planar model 使用专门的快速聚焦近似，所以不保证逐项完全一致。

当前锁定版本在 Windows Julia 1.12.6 上对完整 CESR 调用 twiss 时，ForwardDiff/GTPSA 进入 BeamTracking implicit_step! 会触发底层 illegal-instruction compiler fault。单元素普通 tracking、有限差分 map 和完整 ring 构造均正常；因此本 notebook 暂用 FiniteDiff 显示 map。这个兼容性问题需要单独向 BeamTracking/SciBmad 上游报告。
